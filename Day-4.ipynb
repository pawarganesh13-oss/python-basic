```python
 a[-1:-4]  #     m   y       n   a  m   e      i  s      m  l  t  r
           #   -15 -14     -12 -11 -10 -9     -7 -6     -4 -3 -2 -1
           #  Start from -1  = r
           #  Stop before -4 = m
           #  Step default   = +1
           # But +1 moves left to right.
           # From r, Python cannot move forward to m, because m is on the left side.  So output is: ''
```

```python
a[-1:-4:-1]   #a[-1:-4] is empty because Python slice moves forward by default. To move backward, use a[-1:-4:-1].
```

```python
a[-4:-1]     # L → R
            # m  y     n  a  m  e     i  s     m  l  t  r
            #                                 -4 -3 -2 -1
```

```python
a[-1:-4:-1]  # R → L
             # m  y     n  a  m  e     i  s     m  l  t  r
             #                                 -4 -3 -2 -1
```

```python
a[0:-10:-1]  #Now step is -1, so Python travels Right to Left.
            #But start index 0 is already the first character.
            #m  y  ...
            #0  1
            #From index 0, there is nothing on the left side.
            #So Python cannot travel backward, and output is empty:
```
