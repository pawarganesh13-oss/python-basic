# FOR LOOP

#A for loop is used to execute a block of code repeatedly for each item in a sequence, such as a list, tuple, string, or range.

#for → Keyword used to start a loop.   
#variable → Stores the current element during each iteration.   
#in → Used to iterate over the sequence.     
#sequence → List, tuple, string, range, etc.    
#: → Indicates the beginning of the loop body.       
#Indentation → Defines the statements inside the loop.

```python
#Q-Q. Write a Python program to add 2 to each element of a given list and store the updated
#values in a new list using a for loop and the append() method.

l=[2,3,4,5.6]
l1=[]
for i in l:
    print(i+2)
    l1.append(i+2)
```

```python
#Q-Q. Write a Python program to extract only integer (int) elements from a list using a for loop and if condition.

l=[1,2,3,56,67,78,'sreenivas',3+4j,[23,34,56,7657454,5456578547436]]
l1=[]
for i in l:
    if type(i)==int:
        l1.append(i)
    print(i)
```

```python
#Q. Write a Python program to extract and print all integer elements from a list, including integers present 
#inside a nested list, using for loops and if-elif conditions.
l=[1,2,3,56,67,78,'sreenivas',3+4j,[23,34,56,7657454,5456578547436]]
l1=[]
for i in l:
    if type(i)==int:
        print(i)
    elif type(i)==list:
        for j in i:
            if type (j)==int:
                print(j)
                l1.append(j)
```

```python
#The range() function generates a sequence of integers within a specified range. It is commonly used in for loops.
#range(start, stop, step)
#start: Starting number (included).
#stop: Ending number (excluded).
#step: Difference between consecutive numbers.
#Important: The stop value is never included.
#type of range
#range(stop)->for i in range(5):     print(i)
#range(start, stop)->for i in range(2, 8):   print(i)
#range(start, stop, step)->for i in range(2, 11, 2):   print(i)
#Negative Step in range()->for i in range(10, 0, -1):  print(i)
```

```python
#enumerate() is a built-in Python function used to get both the index number and value of each element while iterating through a sequence.
#i → Index number
#j → Actual element value
#enumerate(l) → Produces index-value pairs.
l = [10, "Python", 20, 4.5, "SQL", 30]
#Without enumerate():
for i in l:
    print(i)
#Here, i represents the value.
#With enumerate():
for i, j in enumerate(l):
    print(i, j)
#Change Starting Index
l = ["SQL", "Python", "ADF"]

for i, j in enumerate(l, start=1):
    print(i, j)
```
