```python
{[3,4,5,6],2,3,4,5}
#Why does this error occur?
#Set {} can contain only hashable elements.
#List [] is mutable (can be changed).
#Lists are unhashable, so we cannot store a list directly inside a set.
```

```python
#Easy Way to Remember
#List → Ordered + Mutable + Duplicates->[]
#Tuple → Ordered + Immutable + Duplicates->()
#Set → Unordered + Mutable + Unique values->{}
```

```python
{(34,56,67),3,4,5}
#Why is this valid?
#Set {} can contain hashable elements.
#Tuple () is immutable.
#The tuple (34, 56, 67) contains only integers, which are hashable.
#Therefore, a tuple can be stored inside a set
```
